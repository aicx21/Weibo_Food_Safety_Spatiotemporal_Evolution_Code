```python
import pandas as pd

# ✅ 1. 读取数据
df = pd.read_excel("你的原始文件.xlsx")  # 替换为你的路径
df.columns = ['时间', '省份', 'label']
df = df.dropna(subset=['label', '省份'])

# ✅ 2. 定义标准省份列表（31个大陆省级行政区，可自行补充港澳台）
province_list = [
    '北京', '天津', '上海', '重庆', '河北', '山西', '辽宁', '吉林', '黑龙江',
    '江苏', '浙江', '安徽', '福建', '江西', '山东', '河南', '湖北', '湖南',
    '广东', '海南', '四川', '贵州', '云南', '陕西', '甘肃', '青海',
    '内蒙古', '广西', '西藏', '宁夏', '新疆'
]

# ✅ 3. 自动识别“发布地址”中包含的省份
def identify_province(addr):
    for p in province_list:
        if p in str(addr):  # 避免 NaN
            return p
    return '其他'  # 如果都匹配不到

df['省份标准化'] = df['省份'].apply(identify_province)

# ✅ 4. 转换 label 为情绪得分：0→-1，1→1，2→0
label_map = {0: -1, 1: 1, 2: 0}
df['情绪值'] = df['label'].map(label_map)

# ✅ 5. 分组聚合
result = df.groupby('省份标准化').agg(
    数量=('label', 'count'),
    情绪得分=('情绪值', 'mean')
).reset_index()

# ✅ 6. 导出结果
result.to_excel("自动识别省份_情绪统计.xlsx", index=False)

print("✅ 自动识别省份完成，结果已保存为 '自动识别省份_情绪统计.xlsx'")
```
